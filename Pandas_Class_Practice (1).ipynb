```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score

np.random.seed(42)

# -----------------------------
# 1. Generate Synthetic Data
# -----------------------------
n_households = 500
hours = 24

data = []

for h in range(n_households):
    base = np.random.uniform(0.5, 2.5)  # base consumption
    responsiveness = np.random.rand()   # behavior factor
    
    for hour in range(hours):
        temp = np.random.uniform(20, 40)
        price = 5 + (hour in range(18, 22)) * 5  # peak price
        
        consumption = base + 0.05*temp + np.random.normal(0, 0.2)
        
        # Peak label
        peak = 1 if hour in range(18, 22) else 0
        
        # Response behavior
        response = 1 if (peak and responsiveness > 0.6) else 0
        
        data.append([h, hour, temp, price, consumption, peak, response])

df = pd.DataFrame(data, columns=[
    "household", "hour", "temp", "price", "consumption", "peak", "response"
])

print(df.head())

# -----------------------------
# 2. Peak Demand Prediction
# -----------------------------
X = df[["temp", "price", "hour"]]
y = df["peak"]

model = RandomForestClassifier()
model.fit(X, y)

pred = model.predict(X)

print("Peak Prediction Accuracy:", accuracy_score(y, pred))

# -----------------------------
# 3. Responsive Household Detection
# -----------------------------
X_resp = df[["consumption", "price", "temp"]]
y_resp = df["response"]

model_resp = RandomForestClassifier()
model_resp.fit(X_resp, y_resp)

df["pred_response"] = model_resp.predict(X_resp)

# -----------------------------
# 4. Clustering (Flexibility Groups)
# -----------------------------
agg = df.groupby("household")[["consumption", "response"]].mean()

kmeans = KMeans(n_clusters=2, random_state=42)
agg["cluster"] = kmeans.fit_predict(agg)

print("\nCluster Summary:\n", agg.groupby("cluster").mean())

# -----------------------------
# 5. Token Incentive Mechanism
# -----------------------------
def token_reward(consumption, peak):
    if peak:
        return max(0, 1.5 - consumption)  # reward lower usage
    return 0

df["tokens"] = df.apply(lambda x: token_reward(x["consumption"], x["peak"]), axis=1)

# -----------------------------
# 6. Demand Reduction Simulation
# -----------------------------
df["adjusted_consumption"] = df["consumption"] - df["tokens"] * 0.2

original = df["consumption"].sum()
reduced = df["adjusted_consumption"].sum()

reduction = (original - reduced) / original * 100

print("\nTotal Reduction (%):", reduction)

# -----------------------------
# 7. Visualization
# -----------------------------
plt.figure(figsize=(8,5))
plt.plot(df.groupby("hour")["consumption"].mean(), label="Original")
plt.plot(df.groupby("hour")["adjusted_consumption"].mean(), label="After DR")
plt.legend()
plt.title("Demand Response Effect")
plt.xlabel("Hour")
plt.ylabel("Consumption")
plt.show()
```
