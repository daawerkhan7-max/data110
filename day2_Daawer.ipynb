```python
#nested if statement
num = int(input("enter a number"))
if num > 0:
    print("number is positive")
    if num % 2 == 0:
        print("number is even")
    else:
            print("number is odd")
elif num < 0:
    print("number is negative")
else:print("the number is Zero")
```

```python
# Break ststement unsing loop

for letters in "MachineLearning":
    if letters == "L":
        break

    print(letters)
print("the loop ends here")
print ("this is break statement")
```

```python
# pass ststement unsing loop

for letters in "MachineLearning":
    if letters == "L":
        pass

    print(letters)
print("the loop ends here")
print ("this is pass statement")
```

```python
# continue ststement unsing loop

for letters in "MachineLearning":
    if letters == 'L':
        continue

    print(letters)
print("the loop ends here")
print ("this is continue statement")
```
