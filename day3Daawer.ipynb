```python
#NESTED LOOP , ZIP FUNCTION, LIST COMPRIHENSHION NUMPY
# WRITE A PROGRAM TO WRITE MULTIPLICATION TABLE OF GIEN NUMBER

num = int (input("enter the number"))
for i in range (1,11):
    result = num*i
    print(num,"*",i,"=",result)
```

```python
#write the multiplication from 2 to 10 using nested loop

for num in range (1,11):
  print ("multiplication table of",num)
  for i in range (1,11):
    result = num*i
    print(num,"*",i,"=",result)
```

```python
#zip function
players = ["virat","aiyer","rajat","david","kronel"]
runs = [120,100,130,199,123]
age = [32,33,35,25,28]
Today_Score =[10,35,100,101,120]
Var_mapped =zip(players,runs,age,Today_Score)
print(list(Var_mapped))
```

```python
#zip function
players = ["virat","venkatesh","rajat","tim","kronel"]
runs = [12,10,13,19,3]
age = [32,33,35,25,28]
Today_Score =[15,30,10,156,110]
Var_mapped = zip(players,runs,age,Today_Score)

P, R, A, T = zip(*Var_mapped)
print(P)
print(R)
print(A)
print(T)
```

```python
pairs = ((1,"a"),(2,"b"),(3,"c"))
numbers,letters = zip(*pairs)
print(numbers)
print(letters)
```

```python
#2d list
planets =[["murcury","venus","earth"],["jupiter","mars","saturn"],["uranus","neptune","pluto"]]
planets
```

```python
single_list =[]
for sublist in planets:
  for single_planet in sublist:
    single_list.append(single_planet)
print(single_list)
```
