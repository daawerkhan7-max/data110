```python
import pandas as pd
import numpy as np

# Sample Dataset
data = {
    "Age": [20, 21, np.nan, 23, 24, np.nan, 26],
    "Marks": [85, np.nan, 78, 90, np.nan, 88, 95],
    "City": ["Delhi", "Mumbai", np.nan, "Delhi", "Chennai", "Mumbai", np.nan]
}

df = pd.DataFrame(data)

print("Original Dataset\n")
print(df)
```

```python
df_drop_rows = df.dropna()

print("\nRows after dropping missing values")
print(df_drop_rows)
```

```python
df_drop_cols = df.dropna(axis=1)

print("\nColumns after dropping missing values")
print(df_drop_cols)
```

```python
df_constant = df.fillna(0)

print("\nFill with Constant (0)")
print(df_constant)
```

```python
df_constant_cat = df.fillna("Unknown")

print(df_constant_cat)
```

```python
df_median = df.copy()

df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())
df_median["Marks"] = df_median["Marks"].fillna(df_median["Marks"].median())

print("\nMedian Imputation")
print(df_median)
```

```python
df_mode = df.copy()

df_mode["City"] = df_mode["City"].fillna(df_mode["City"].mode()[0])

print("\nMode Imputation")
print(df_mode)
```

```python
df_ffill = df.ffill()

print("\nForward Fill")
print(df_ffill)
```

```python
# Backward Fill
df_bfill = df.bfill()

print("\nBackward Fill")
print(df_bfill)
```

```python
df_interp = df.copy()

df_interp["Age"] = df_interp["Age"].interpolate()
df_interp["Marks"] = df_interp["Marks"].interpolate()

print("\nLinear Interpolation")
print(df_interp)
```

```python
df_specific = df.fillna({
    "Age":18,
    "Marks":50,
    "City":"Hyderabad"
})

print("\nSpecific Value Imputation")
print(df_specific)
```

```python
from sklearn.impute import KNNImputer

knn = KNNImputer(n_neighbors=3)

df_knn = df[["Age","Marks"]]

df_knn = pd.DataFrame(
    knn.fit_transform(df_knn),
    columns=df_knn.columns
)

print("\nKNN Imputation")
print(df_knn)
```

```python
from sklearn.impute import SimpleImputer

imp = SimpleImputer(strategy="mean")

df_mean_sk = pd.DataFrame(
    imp.fit_transform(df[["Age","Marks"]]),
    columns=["Age","Marks"]
)

print("\nMean Imputation (Scikit-learn)")
print(df_mean_sk)
```
